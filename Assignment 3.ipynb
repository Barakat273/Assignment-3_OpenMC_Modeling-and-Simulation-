```python
import openmc
```

# 1. Define Monoenergetic Isotropic Point Sources
Photons with an energy of 1*10^6 eV.

Neutrons with an energy of 2.5 * 10^−3 eV.

```python
# Monoenergtic photons source #

photon_point_source = openmc.IndependentSource()
photon_point_source.particle = 'photon'
photon_point_source.space = openmc.stats.Point((0.0, 0.0, 0.0))
photon_point_source.angle = openmc.stats.Isotropic()
photon_point_source.energy = openmc.stats.Discrete([1.0e6], [1.0])  # 1 MeV photons

openmc.settings.source = photon_point_source
```

```python
# Monoenergtic neutrons source #

neutron_point_source = openmc.IndependentSource()
neutron_point_source.particle = 'neutron'
neutron_point_source.space = openmc.stats.Point((0.0, 0.0, 0.0))
neutron_point_source.angle = openmc.stats.Isotropic()
neutron_point_source.energy = openmc.stats.Discrete([2.5e-3], [1.0])  # 0.0025 eV neutrons

openmc.settings.source = neutron_point_source
```

# 2. Define a Monodirectional Photon Point Source

```python
# Unidirectional photon source #

mono_photon_source = openmc.IndependentSource()
mono_photon_source.particle = 'photon'
mono_photon_source.space = openmc.stats.Point((0.0, 0.0, 0.0))
mono_photon_source.angle = openmc.stats.Monodirectional((1, 0, 0))  # Fixed direction (x-axis)
mono_photon_source.energy = openmc.stats.Discrete([1.0e6, 2.0e6], [0.5, 0.5])  # 1 MeV and 2 MeV photons with probabilities 0.5 and 0.5 

openmc.settings.source = mono_photon_source
```

# Group of Sources 

```python
all_sources = [photon_point_source, neutron_point_source, mono_photon_source]
```
